Run state-of-the-art **AI-generated image & video detectors** in a few lines with the [`agid`](https://github.com/SMSD75/AGID) library.

> Tip: switch the runtime to a **GPU** (Runtime → Change runtime type) — the backbones are large.

## 1. Install

Installs `agid` (which pulls PyTorch, transformers, etc.) and OpenAI CLIP (needed by a couple of detectors). Takes a few minutes on first run.

```python
%pip install -q "git+https://github.com/SMSD75/AGID.git" "git+https://github.com/openai/CLIP.git"
```

```python
import os
# Point at the model zoo so weights auto-download. Change to your own repo if needed.
os.environ.setdefault("AGID_HUB_REPO", "SMSD75/agid-checkpoints")

import agid
agid.list_models()  # detectors + which ones can run
```

## 2. Upload an image (or video)

```python
try:
    from google.colab import files
    uploaded = files.upload()
    path = next(iter(uploaded))
except Exception:
    path = "your_image.png"  # running outside Colab: set a local path
print("file:", path)
```

## 3. Detect

Weights download automatically on first use. Try other detectors: `cnn`, `co_spy`, `dinov2`, or `d3` for video.

```python
det = agid.load("clip")
result = det.predict(path)
print(f"Verdict: {result['verdict'].upper()}  |  P(AI-generated) = {result['fake_probability']:.3f}")
result
```

## Next steps

- Compare several detectors on the same file: `for k in ['clip','cnn','co_spy']: print(k, agid.load(k).predict(path)['fake_probability'])`
- Use your own trained checkpoint: `agid.load('cnn', checkpoint='/path/to.pth')`
